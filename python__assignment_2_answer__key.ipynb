ans.9  Break, continue and pass are control flow statement used within loops to manage execution flow. Break :- The break statement is used to exit from a loop prematurely. It can be used in for loops or while loops. When a break statement is encountered, the loop is terminated. Continue :- The continue statement is used to skip the remaining code inside the current iteration of the loop and proceed to the next iteration. The loop to jump to the next iteration without executing the rest of the code in the loop body for the current iteration.
pass :- The pass statement is a null operation. It is used as a placeholder for future code. It does nothing and is often used in places where code is syntactically required. It can be used in loops, functions, classes, or conditional blocks where you need a statement but don't want to add any functionality.
